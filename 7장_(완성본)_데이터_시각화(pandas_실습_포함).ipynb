### 히트맵 - matshow
> 히트맵은 데이터의 분포나 패턴을 색상을 이용하여 표현하는 시각화 방법이다. 데이터의 크기나 빈도에 따라 그 값이 클수록 진한 색상으로 표현하며, 이는 전체적인 분포를 직관적으로 보기 쉽게 만든다. 주로 2차원이나 3차원 분포에서 각 위치의 값을 색상으로 표시하는 방식으로 활용된다.
여러 개의 변수 간의 상관 관계를 분석할 때, 히트맵을 사용하면 상관 계수들 간의 분포를 시각적으로 쉽게 이해할 수 있다. 특히 상관 계수는 범위가 -1에서 1로 제한되어 있기 때문에 이를 히트맵으로 나타내면 더 직관적인 결과를 제공한다. 상관 계수가 -1에 가까운 변수들과 1에 가까운 변수들이 대조적으로 나타나며, 색의 진하기에 따라 상관 계수의 크기를 쉽게 비교할 수 있다.
[네이버 지식백과] 히트맵 [Heatmap] (두산백과 두피디아, 두산백과)


```python
import matplotlib.pyplot as plt
import numpy as np

x=np.arange(1,5)
y=np.arange(2,6)
arr=(x,y)
plt.matshow(arr)
plt.colorbar()
plt.show()
```

sklearn을 이용하여 iris Data 불러오기

```python
from sklearn.datasets import load_iris
iris=load_iris()
iris
```

```python
import pandas as pd
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['label']=iris.target
df.head()
```

```python
# 꽃받침 길이와 꽃잎 길이 분포 확인(품종별 색상 다르게 표시)
import matplotlib.pyplot as plt

x = df['sepal length (cm)']
y = df['petal length (cm)']
plt.scatter(x,y,c=df['label'])
plt.xlabel('sepal length (cm)',size=15)
plt.ylabel('petal length (cm)',size=15)
plt.show()
```

```python
import seaborn as sns
sns.set()
sns.pairplot(data=df, hue='label')
plt.show()
```
